# Figure 5: per-session median spike features by organoid class

Panels 5A-5D show, for each class, one point per session: the median of that session's spikes for Amplitude, Peak A, Channel ISI and Trough to Peak. Every class is tested against hCO^APP (hCO is tested against hCO^APP) with a two-tailed Mann-Whitney U test on the signed medians.

- Unsorted spikes from `input/unsorted` (hCO, hCO^APP) and `input/drug_test` (MK-8722, hCOL^APP). All 89 sessions are used, with no minimum-spike or DNU filtering.
- The two MK-8722 pickles store `Amplitude (uV)` as a positive magnitude, so their signed amplitude is `Trough (uV) - Peak B (uV)`, as in Table 1. Amplitude is plotted as a magnitude and sessions with a positive median are drawn in red.
- Point jitter is seeded, so the plotted x positions are reproducible and saved in the CSVs.

```python
from manuscript_analysis.figures.feature_medians import (
    PANELS, export_feature_boxplot, load_session_medians, mann_whitney_tests,
)
from manuscript_analysis.figures.fig_utils import get_out_path, report_saved

session_medians = load_session_medians()
tests = mann_whitney_tests(session_medians)

saved_paths = []
for panel in PANELS:
    _, panel_paths = export_feature_boxplot(
        session_medians, tests, panel, get_out_path(panel.figure_id)
    )
    saved_paths += panel_paths
report_saved(saved_paths)

session_medians.groupby('organoid_class', sort=False)['session'].nunique()
```
