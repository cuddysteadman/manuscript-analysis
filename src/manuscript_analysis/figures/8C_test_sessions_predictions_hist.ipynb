# Figure 8C: insertion-session prediction-score histograms

```python
import pandas as pd

from manuscript_analysis.figures.fig_utils import RECORDING_SCORE_EXPORTS, get_out_path, report_saved
from manuscript_analysis.figures.score_hist import export_panel_histograms, split_panels

PANELS = {
    'a': 'AD-CO-DAY150',
    'b': 'AD-CO-DAY150-MKAD2-77-10NM-30DAYS',
    'c': 'AD-CO-DAY150-MKAD2-77-100NM-30DAYS',
}

scores = (
    pd.read_csv(RECORDING_SCORE_EXPORTS / 'insertion_model_recording_scores.csv')
    .rename(columns={'session': 'recording', 'class_name': 'condition'})
    .drop(columns='optimal_threshold')
)
panel_data = split_panels(scores, PANELS)
pd.concat(panel_data.values()).groupby(['panel', 'condition'], sort=False).size().reset_index(name='n_prediction_scores')
```

```python
report_saved(export_panel_histograms(panel_data, get_out_path('8C'), '8C_test_sessions_predictions_hist'))
```
