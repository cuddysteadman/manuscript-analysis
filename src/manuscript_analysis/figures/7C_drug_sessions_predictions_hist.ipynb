# Figure 7C: drug-session prediction-score histograms

```python
import pandas as pd

from manuscript_analysis.contract import N_RUNS
from manuscript_analysis.figures.fig_utils import RECORDING_SCORE_EXPORTS, get_out_path, report_saved
from manuscript_analysis.figures.score_hist import export_panel_histograms, split_panels

PANELS = {
    'a': 'AD-CO-DAY150-MKAD2-77-10NM-30DAYS',
    'b': 'AD-CO-DAY150-MKAD2-77-100NM-30DAYS',
    'c': 'OL-AD-CO-DAY150',
}

raw = pd.read_csv(RECORDING_SCORE_EXPORTS / 'drug_test_model_recording_scores.csv')
scores = pd.DataFrame({
    'seed': raw['model_id'].str.removeprefix('seed').astype(int),
    'session': raw['recording'].str.removesuffix('.dat'),
    'condition': raw['condition'],
    'probability_ad': raw['probability_ad'],
}).sort_values(['seed', 'session'], kind='stable')
assert scores['seed'].nunique() == N_RUNS and not scores.duplicated(['seed', 'session']).any()

panel_data = split_panels(scores, PANELS)
(
    pd.concat(panel_data.values())
    .groupby(['panel', 'condition'], sort=False)['session']
    .agg(n_sessions='nunique', n_prediction_scores='size')
    .reset_index()
)
```

```python
report_saved(export_panel_histograms(panel_data, get_out_path('7C'), '7C_drug_sessions_predictions_hist'))
```
