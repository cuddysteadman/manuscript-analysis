# UNSORTED null distribution

```python
from manuscript_analysis import find_project_root
from manuscript_analysis.workflows import null

ROOT = find_project_root()
PROFILE = "SMOKE"
N_PERMUTATIONS = 100
PROFILES = {"SMOKE": (range(2), range(2)), "FULL": (range(N_PERMUTATIONS), range(25))}
PERMUTATIONS, SPLITS = PROFILES[PROFILE]
OUTPUT_DIR = ROOT / "results" / "null-distribution"
```

## Run or resume

```python
STATE = null.setup(
    ROOT / "input", OUTPUT_DIR,
    permutation_seeds=PERMUTATIONS, split_seeds=SPLITS,
)
try:
    COMPLETED = null.run_folds(
        STATE, permutation_seeds=PERMUTATIONS, split_seeds=SPLITS
    )
    RESULTS = null.consolidate(
        OUTPUT_DIR, permutation_seeds=PERMUTATIONS, split_seeds=SPLITS
    )
finally:
    null.cleanup(STATE)
RESULTS["null_distribution"]
```
